# Balance Sheet Analysis

This notebook computes key balance sheet metrics such as total assets, net debt,
and capital structure ratios using quarterly financial data.

```python
from src.setup import *
```

## Step 1: Define Balance Sheet Analysis Function

This function calculates key balance sheet metrics based on the latest available
quarterly financial data and the company’s balance sheet.

```python
def balance_sheet_analysis(quarterly_balance_sheet):
    """
    Compute key balance sheet metrics using latest quarterly data.

    Parameters
    ----------
    quarterly_balance_sheet : pandas.DataFrame
        Quarterly statement snapshot with the latest period first.

    Returns
    -------
    dict
        Dictionary with total assets, debt, equity, cash, net debt, and ratios.
    """
    total_assets = quarterly_balance_sheet.loc["TotalAssets"].iloc[0]
    total_debt = quarterly_balance_sheet.loc["TotalDebt"].iloc[0]
    equity = quarterly_balance_sheet.loc["StockholdersEquity"].iloc[0]
    cash = quarterly_balance_sheet.loc["CashAndCashEquivalents"].iloc[0]
    long_term_debt = quarterly_balance_sheet.loc["LongTermDebt"].iloc[0]

    net_debt = total_debt - cash
    gearing_ratio = net_debt / equity if equity != 0 else np.nan
    equity_ratio = equity / total_assets if total_assets != 0 else np.nan
    debt_ratio = total_debt / total_assets if total_assets != 0 else np.nan

    return {
        "Total Assets": total_assets,
        "Total Debt": total_debt,
        "Equity": equity,
        "Cash": cash,
        "Long Term Debt": long_term_debt,
        "Net Debt": net_debt,
        "Gearing Ratio": gearing_ratio,
        "Equity Ratio": equity_ratio,
        "Debt Ratio": debt_ratio,
    }
```

```python
balance_results = balance_sheet_analysis(quarterly_balance_sheet)
```

## Step 3: Display Results

The results are formatted into a clean table for easier interpretation.

```python
print(f"Balance sheet date: {quarterly_balance_sheet.columns[0].date()}")
df_display = pd.DataFrame(balance_results, index=[0]).T.rename(columns={0: "Value"})
with pd.option_context("display.float_format", "{:,.2f}".format):
    display(df_display)
```
